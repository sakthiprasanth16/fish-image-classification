Image Count

```python
import os

data_dir = "data"
folders = ['train', 'val', 'test']

for split in folders:
    split_path = os.path.join(data_dir, split)
    print(f"\n📁 {split.upper()} FOLDER:")

    for class_name in os.listdir(split_path):
        class_path = os.path.join(split_path, class_name)
        if os.path.isdir(class_path):
            num_images = len([
                f for f in os.listdir(class_path)
                if os.path.isfile(os.path.join(class_path, f)) and f.lower().endswith(('.jpg', '.jpeg', '.png'))
            ])
            print(f"  🔸 {class_name}: {num_images} images")
```

Data Augmentation for animal fish bass train,test,val data

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator, load_img, img_to_array
import os

# Directory settings
augment_dir = "data/train/animal fish bass"
save_dir = augment_dir  # Save in the same directory

# Target number of images you want in total
target_total = 564

# Data augmentation configuration
datagen = ImageDataGenerator(
    rotation_range=30,
    width_shift_range=0.2,
    height_shift_range=0.2,
    zoom_range=0.2,
    shear_range=0.2,
    horizontal_flip=True,
    fill_mode='nearest'
)

# Get list of valid image files
valid_extensions = ('.jpg', '.jpeg', '.png')
existing_images = [f for f in os.listdir(augment_dir) if f.lower().endswith(valid_extensions)]
current_count = len(existing_images)
to_generate = target_total - current_count

if to_generate <= 0:
    print(f"✅ Already have {current_count} images. No augmentation needed.")
    exit()

print(f"🔄 Found {current_count} images. Generating {to_generate} more to reach {target_total}...")

# Augmentation loop
generated = 0
img_index = 0

while generated < to_generate:
    img_file = existing_images[img_index % current_count]  # Cycle through all available images
    img_path = os.path.join(augment_dir, img_file)

    img = load_img(img_path, target_size=(256, 256))
    x = img_to_array(img)
    x = x.reshape((1,) + x.shape)

    # Generate 1 new image from the current image
    for batch in datagen.flow(x, batch_size=1, save_to_dir=save_dir, save_prefix='aug', save_format='jpg'):
        generated += 1
        break  # One image per iteration

    img_index += 1  # Move to next image

print(f"✅ Augmentation complete. Total images now: {current_count + generated}")
```
